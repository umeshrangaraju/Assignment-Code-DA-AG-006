# **01. What are Type I and Type II errors in hypothesis testing, and how do they impact decision-making ?**

Type I and Type II Errors in Hypothesis Testing

In statistical hypothesis testing, decisions are made about population parameters based on sample data. Since these decisions are probabilistic, two kinds of errors can occur: Type I and Type II errors. Understanding these errors is essential because they influence how tests are designed, how significance levels are chosen, and how results are interpreted.

1. Type I Error (False Positive)

A Type I error occurs when the researcher rejects a true null hypothesis (H₀).

Symbol: α (alpha)  
Also called the “false alarm” error.

Example:

Suppose a pharmaceutical company tests a new drug that in reality has no effect. If the statistical test concludes that the drug does work, the company has made a Type I error.

Interpretation:

You are detecting an effect or difference that does not actually exist.

Consequences:

Approving an ineffective drug or medical treatment     
Concluding a policy works when it does not   
Wasting resources on false discoveries

Because Type I errors can have serious implications, researchers often set α = 0.05 or 0.01 to limit the probability of this mistake.

2. Type II Error (False Negative)

A Type II error occurs when the researcher fails to reject a false null hypothesis.

Symbol: β (beta)     
Also called the “missed detection” error.

Example:

Using the drug example, if the new drug actually does work but the test    concludes that it does not, the researcher has made a Type II error.

Interpretation:

You fail to detect an effect or difference that actually exists.

Consequences:

Missing out on a beneficial treatment or innovation    
Ignoring a real safety hazard    
Underestimating relationships or differences in scientific studies

3. Relationship Between Type I and Type II Errors

There is a trade-off between the two errors.

Lowering α (to reduce Type I errors) often increases β (more Type II errors).  
Increasing sample size can reduce both errors simultaneously.

Thus, careful planning—often through power analysis—is needed to balance these risks.

4. Impact on Decision-Making

These errors influence several steps in the research process:

a) Choice of Significance Level (α)

Decision-makers choose α depending on the seriousness of Type I error.   
In medicine or aviation safety: α is kept very low (e.g., 0.01).   
In exploratory research: higher α may be acceptable.

b) Determining Sample Size

To reduce Type II errors, researchers increase sample size, ensuring higher statistical power.

c) Policy and Business Decisions

Type I error may lead to adopting ineffective policies or products.    
Type II error may lead to rejecting beneficial innovations or failing to detect fraud/risk.

d) Risk Management

Organizations evaluate which type of error is more costly and design their testing strategy accordingly.

5.Real-World Examples

Medical Testing:

Type I: Diagnosing a healthy person as sick → anxiety, unnecessary treatment.  
Type II: Failing to detect a real disease → delay in treatment.

Quality Control:

Type I: Rejecting a good batch of products → increased cost.         
Type II: Accepting a faulty batch → safety issues, recalls.

Conclusion

Type I and Type II errors are central concepts in hypothesis testing.

Type I error (α): Rejecting a true null hypothesis (false positive).   
Type II error (β): Failing to reject a false null hypothesis (false negative).

These errors shape how significance levels are chosen, how tests are designed, and how decisions are made in scientific, business, medical, and policy contexts. Proper understanding allows researchers to balance risks and make more informed, reliable decisions.



## **Question 2: What is the P-value in hypothesis testing, and how should it be interpreted in the context of the null hypothesis ?**

1. Definition of the P-value

The P-value is the probability of obtaining a test statistic as extreme as, or more extreme than, the one observed in the sample, assuming that the null hypothesis (H₀) is true.

It measures the degree of compatibility between the sample data and the null hypothesis.

2. Role of the P-value in hypothesis testing

In hypothesis testing, we compare the P-value to a pre-selected significance level (α), usually 0.05.

The P-value helps determine whether the observed results provide enough evidence to reject H₀.   
It quantifies the strength of evidence against the null hypothesis.

3.Interpretation of the P-value

(a) When P-value is small (P ≤ α)

  A small P-value indicates that the observed data would be very unlikely if the null hypothesis were true.  
  This provides strong evidence against H₀.  
  Therefore, we reject the null hypothesis.    
  Conclusion: The results are statistically significant.

(b) When P-value is large (P > α)

  A large P-value indicates that the observed data are consistent with the null hypothesis.
  There is insufficient evidence to reject H₀.   
  We fail to reject the null hypothesis, but we do not prove that H₀ is true.  
  Conclusion: The results are not statistically significant.

(c) What the P-value does not mean

To demonstrate deeper understanding (important for high marks):

It is not the probability that the null hypothesis is true.      
It is not the probability that the alternative hypothesis is false.    
It does not measure the size or importance of an effect.     
It only measures the strength of evidence in the sample against H₀.    

4. Example

Suppose a researcher wants to test if a new drug changes average blood pressure.

Step 1: Set hypotheses

  H₀: The drug has no effect on mean blood pressure.    
  H₁: The drug has an effect on mean blood pressure.

Step 2: Conduct test

  After collecting sample data, the calculated P-value is 0.02.

Step 3: Compare P-value with α (0.05)

  Since 0.02 < 0.05, the P-value is small.

Step 4: Interpretation

There is only a 2% chance of getting results as extreme as those observed if the null hypothesis were true.    
These results are unlikely under H₀ → strong evidence against H₀.    
Therefore, the researcher rejects the null hypothesis and concludes that the drug likely has a significant effect on blood pressure.








# **Question 3: Explain the difference between a Z-test and a T-test, including when to use each.**

Introduction

Both the Z-test and the T-test are statistical hypothesis tests used to determine whether there is a significant difference between sample statistics and population parameters. While they are conceptually similar, they differ in the conditions under which they are used, particularly regarding sample size and knowledge of population variance.

1. Z-Test   
Definition  
A Z-test is a statistical test used to determine whether the mean of a sample differs significantly from a known population mean when certain conditions are met.

When to Use a Z-Test    
A Z-test is appropriate when:

The population variance (σ²) or standard deviation (σ) is known.   
The sample size is large (n ≥ 30), making the sampling distribution approximately normal due to the Central Limit Theorem.   
The data follows (or approximately follows) a normal distribution.

Example of Z-Test   
A company claims that the average lifetime of its light bulbs is 1000 hours with a known standard deviation of 100 hours. A sample of 50 bulbs has a mean lifetime of 980 hours.    
Since the standard deviation is known and n = 50 ≥ 30, a Z-test is appropriate to check if the bulbs’ lifetime is significantly lower than claimed.

2. T-Test     
Definition    
A T-test is a statistical test used to determine whether the mean of one or two samples significantly differs from a population mean or another sample mean when the population standard deviation is unknown.

When to Use a T-Test  
A T-test is used when:    
The population variance or standard deviation is unknown.    
Sample size is small (n < 30).           
Data is from a normally distributed population.
Sample variability must be estimated using the sample standard deviation (s).

Types of T-Tests
One-sample T-test – Compare sample mean to a known population mean.
Independent two-sample T-test – Compare means of two independent groups.
Paired sample T-test – Compare means from the same group at two different times.

Example of T-Test

A nutritionist wants to test whether a new diet plan affects weight. She measures the weights of 15 participants before and after the program.
Since the sample size is small (n < 30) and population standard deviation is unknown, a paired T-test is appropriate.

Another example:
A teacher believes that the average exam score of her class is different from the national average of 70. She takes a sample of 20 students and calculates a mean of 75, but the population standard deviation is unknown.
Here, a one-sample T-test is used.

When the population variance is unknown and the sample is small, using a Z-test would underestimate the true variability.

Conclusion

In summary, both Z-tests and T-tests help test hypotheses about population means, but they differ in the key assumptions of sample size and knowledge of population standard deviation. A Z-test is used for large samples with known population variance, while a T-test is used when the sample size is small and the population variance is unknown. Choosing the correct test ensures accurate and reliable statistical conclusions.




# **Question 4: What is a confidence interval, and how does the margin of error influence its width and interpretation ?**

1. Definition of a Confidence Interval (CI)

A confidence interval is a range of values, derived from sample data, that is likely to contain the true population parameter (such as the mean or proportion) with a specified level of confidence.

It is expressed in the form: Estimate±Margin of Error

For example, a 95% confidence interval for a population mean might be:
50 ± 4 ⇒ ( 46 , 54 )

This means we are 95% confident that the true population mean lies between 46 and 54.

2. Interpretation of a Confidence Interval
A 95% confidence interval does not mean there is a 95% probability that the true value is in the interval.

Instead, it means:    
If we repeated the sampling process many times, about 95% of the confidence 	intervals constructed would contain the true population parameter.

Thus, the CI reflects the precision of our estimate:    
Narrow interval → more precise estimate    
Wide interval → less precise estimate    

3. Margin of Error (MoE)    
The margin of error represents the amount of uncertainty in the estimate. It depends on:

MoE = 	Z × 	σ  / n
                        
Where:    
Z = critical value based on confidence level (e.g., 1.96 for 95%)    
σ = population standard deviation (or sample SD as an estimate)    
n = sample size  

4. How Margin of Error Influences the Width of the CI

(a) Larger Margin of Error → Wider Confidence Interval
A larger margin of error gives a broader range of possible values.
This happens when:

The confidence level increases (e.g., from 90% to 99%)   
Sample size is small   
Data is more variable (higher standard deviation)  

Example:    
Suppose the sample mean = 70.   
With MoE = 3 → CI = (67, 73)   
With MoE = 8 → CI = (62, 78)

A MoE of 8 produces a much wider interval, indicating lower precision.

(b) Smaller Margin of Error → Narrower Confidence Interval
A smaller MoE results from:

Lower confidence level   
Larger sample size     
Less variability in data     

Example:
Sample mean = 50      
MoE = 1 → CI = (49, 51)     
MoE = 5 → CI = (45, 55)   

The first interval is more precise and easier to use in decision-making.

5.How MoE Affects Interpretation of CI    

A. Decision-making precision    
A narrow CI gives a precise estimate, making decisions more reliable.      
A wide CI suggests more uncertainty, requiring cautious interpretation.         

B. Statistical conclusions
If a CI does not include a certain value (e.g., 0 for mean difference), we might 		conclude that the effect is statistically significant.  
A wide CI may include many possibilities, making conclusions weaker.

6. Real-World Example   
  Example: Public opinion poll

A survey reports: Support for Candidate A=52%±4%  
This gives a 95% CI: (48%,56%)    
Interpretation: We are 95% confident that the true support lies between 48% and 56%.    
If the margin of error were ±1%, the CI becomes:(51%,53%)   

This is much narrower and indicates greater precision.

If the margin of error were ±8%, the CI would be:(44%,60%)

This is too wide to make a strong conclusion.

7. Summary    
Concept	    	    			             Explanation

2.Margin of error			Amount of uncertainty around the estimate   
3.Larger MoE				Wider CI → less precision

5.Influencing factors			Confidence level, sample size, variability

Conclusion

A confidence interval provides a range of values that is likely to contain the true population parameter with a specified level of confidence. It conveys both the estimate and the uncertainty around it. The margin of error plays a critical role by determining the width of the interval. A larger margin of error produces a wider confidence interval, reflecting greater uncertainty, while a smaller margin of error yields a narrower interval, indicating more precision. Understanding the interplay between margin of error, width of the interval, and confidence level is essential for accurate statistical interpretation and informed decision-making.



# **Question 5: Describe the purpose and assumptions of an ANOVA test. How does it extend hypothesis testing to more than two groups?**

1. Purpose of ANOVA (Analysis of Variance)
The purpose of ANOVA is to determine whether the means of three or more independent groups are significantly different. Instead of performing several pairwise t-tests—which increases the chance of a Type I error—ANOVA provides one overall statistical test to check for mean differences.

ANOVA looks at how much the group means differ compared to how much individuals differ within each group.

It essentially answers:   
“Are the differences between groups larger than what would be expected by random chance?”

2. Assumptions of ANOVA     
(1) Independence of Observations    
Each data point must be independent of the others.  
Proper sampling or random assignment must be used.    
Violation leads to misleading F-statistics.   

(2) Normality of Residuals  
Scores within each group should follow an approximately normal distribution.
ANOVA is robust if sample sizes are large (Central Limit Theorem), but strong 	skewness may reduce accuracy.

(3) Homogeneity of Variances (Equal Variances)    
All groups should have roughly equal variances.   
Checked using Levene’s or Bartlett’s test.    
If violated, use Welch’s ANOVA.

(4) Random Sampling (Optional but desirable)
The sample should represent the population.

3. How ANOVA Extends Hypothesis Testing to More Than Two Groups
Problem with Using Multiple t-tests   
Example: Comparing 4 teaching methods
You would need:     
A vs B    
A vs C    
A vs D  
B vs C    
B vs D    
C vs D    
= 6 separate t-tests    
This inflates Type I error drastically.   

ANOVA Solution    
ANOVA tests all mean differences simultaneously with one F-test.    

ANOVA Hypotheses    
Null hypothesis (H₀): μ₁ = μ₂ = μ₃ = … = μₖ (all group means are equal)   
Alternative hypothesis (H₁): At least one group mean differs

Mechanism   
ANOVA compares:   
Between-group variance (differences in group means)
Within-group variance (variation inside groups)

Using the formula:  
F 		=	Between-group variance    
Within-group variance   
	​
If F is large, it indicates groups differ significantly.

5. Examples of ANOVA in Real Situations   
Example 1: Education (One-Way ANOVA)      
A researcher wants to compare exam scores across three teaching methods:
Lecture      
Group Discussion     
Online Learning    
If ANOVA finds p < 0.05, it means at least one method leads to different    average scores.    


Example 2: Agriculture  
A farmer tests three types of fertilizers on plant height:      
Fertilizer		Mean Height   
A			15 cm     
B			18 cm       
C			22 cm     

ANOVA determines if fertilizer type significantly affects plant growth.   

Example 3: Medicine   
A doctor compares four pain-relief drugs on recovery time.      
ANOVA checks if the average effectiveness differs among the drugs.    
If significant, post-hoc tests (Tukey, Bonferroni) determine which drugs differ.  

Example 4: Marketing      
A company tests whether three advertisement types produce different mean sales:      
TV ads       
Social media ads     
Print ads        

ANOVA helps determine the most effective advertising strategy.      

6. Conclusion     
ANOVA is a powerful statistical method used to compare three or more group means. It avoids the inflated Type I error rate caused by multiple t-tests by using a single F-test. ANOVA relies on assumptions of independence, normality, and equal variances. By partitioning total variation into between-group and within-group variation, ANOVA determines whether group means are significantly different.

# Question 6: Write a Python program to perform a one-sample Z-test and interpret the result for a given dataset.

### **1. Purpose**

To determine if there is a statistically significant difference between a sample mean and a hypothesized population mean when the population standard deviation is known.



### 2. Assumptions
Independence: Observations are independent of each other.    
Normality: The data is approximately normally distributed, or the sample size is large (n ≥ 30) so the Central Limit Theorem applies.    
Known Population Standard Deviation: The standard deviation of the population (σ) is known.

### 3. Example Scenario: Test Scores

A high school principal claims that the average score on a national standardized test for students in her school is 85. The national average score is known to be 80 with a population standard deviation of 10. A random sample of 40 students from the school took the test, and their average score was 83.    

We want to determine if the principal's claim (that her school's average is 85) is significantly different from the national average of 80, or if the sample mean of 83 provides enough evidence to suggest that her school's students perform differently from the national average.   

```python
import numpy as np
from scipy import stats

# 1. Define Hypotheses
# Null Hypothesis (H0): The school's average score is equal to the national average (μ = 80)
# Alternative Hypothesis (H1): The school's average score is different from the national average (μ ≠ 80)

# Given data
sample_mean = 83        # Sample mean score from the principal's school
population_mean_h0 = 80 # Hypothesized population mean under the null hypothesis (national average)
population_std = 10     # Known population standard deviation
sample_size = 40        # Number of students in the sample
alpha = 0.05            # Significance level

print(f"Sample Mean: {sample_mean}")
print(f"Hypothesized Population Mean (H0): {population_mean_h0}")
print(f"Known Population Standard Deviation: {population_std}")
print(f"Sample Size: {sample_size}")
print(f"Significance Level (alpha): {alpha}")
```

### **4. Calculate the Z-statistic**

The Z-statistic is calculated using the formula:

$Z = \frac{\bar{X} - \mu_0}{\sigma / \sqrt{n}}$

Where:
*   $\bar{X}$ is the sample mean
*   $\mu_0$ is the hypothesized population mean
*   $\sigma$ is the population standard deviation
*   $n$ is the sample size

```python
# Calculate the standard error of the mean (SEM)
sem = population_std / np.sqrt(sample_size)

# Calculate the Z-statistic
z_statistic = (sample_mean - population_mean_h0) / sem

print(f"Standard Error of the Mean (SEM): {sem:.4f}")
print(f"Z-statistic: {z_statistic:.4f}")
```

### **5. Calculate the P-value**

The P-value is the probability of observing a Z-statistic as extreme as, or more extreme than, the one calculated, assuming the null hypothesis is true. Since this is a two-tailed test (because H1 states "different from", not "greater than" or "less than"), we multiply the one-tailed probability by 2.

```python
# Calculate the p-value for a two-tailed test
# stats.norm.sf(z) gives P(Z > z)
p_value = 2 * stats.norm.sf(abs(z_statistic))

print(f"P-value: {p_value:.4f}")
```

### **6. Interpret the Result**

Compare the P-value to the significance level ($\alpha$).

*   If P-value ≤ $\alpha$: Reject the null hypothesis.
*   If P-value > $\alpha$: Fail to reject the null hypothesis.

```python
print(f"Significance Level (alpha): {alpha}")
print(f"P-value: {p_value:.4f}")

if p_value <= alpha:
    print("\nDecision: Reject the Null Hypothesis (H0).")
    print(f"Conclusion: There is sufficient evidence at the {alpha*100}% significance level to conclude that the average test score of students in the principal's school is significantly different from the national average of {population_mean_h0}.")
else:
    print("\nDecision: Fail to Reject the Null Hypothesis (H0).")
    print(f"Conclusion: There is NOT sufficient evidence at the {alpha*100}% significance level to conclude that the average test score of students in the principal's school is significantly different from the national average of {population_mean_h0}. The observed difference could be due to random chance.")
```

### **Summary of the Z-Test Steps:**

1.  **State the Hypotheses**: Formulate H₀ and H₁.
2.  **Set the Significance Level ($\alpha$)**: Typically 0.05.
3.  **Calculate the Z-statistic**: Using the sample data and population parameters.
4.  **Calculate the P-value**: The probability of observing such a result if H₀ were true.
5.  **Make a Decision**: Compare P-value with $\alpha$.
6.  **Formulate a Conclusion**: Interpret the decision in the context of the problem.

# **Question 7:Simulate a dataset from a binomial distribution (n = 10, p = 0.5) using NumPy and plot the histogram.**

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters for the binomial distribution
n = 10  # Number of trials
p = 0.5 # Probability of success
size = 1000 # Number of simulations

# Simulate a dataset from a binomial distribution
binomial_data = np.random.binomial(n, p, size)

print(f"Generated {size} samples from a binomial distribution (n={n}, p={p}).")
print(f"First 10 values: {binomial_data[:10]}")
```

Define a non-normal distribution (Exponential) and its parameters, including the rate parameter (lambda), the number of individual samples to draw, and the size of each sample.

To define the non-normal distribution (Exponential) parameters and sampling details, I will import numpy and set the specified values for lambda_param, sample_size_clt, and num_samples.



```python
import numpy as np

# 1. Define the rate parameter (lambda) for the Exponential distribution
lambda_param = 0.5

# 2. Define the number of individual samples to draw (for CLT demonstration)
sample_size_clt = 50

# 3. Define the total number of samples to draw
num_samples = 1000

print(f"Lambda (rate parameter for Exponential distribution): {lambda_param}")
print(f"Sample size for each CLT sample: {sample_size_clt}")
print(f"Number of samples to draw: {num_samples}")
```

## Generate Samples from Non-Normal Distribution

### Subtask:
Generate 'num_samples' samples from the chosen non-normal distribution (Exponential). Each sample should consist of 'sample_size_clt' data points.


**Reasoning**:
To generate multiple samples from the Exponential distribution as per the subtask's instructions, I will use `numpy.random.exponential` with the previously defined parameters for rate, sample size, and number of samples.



```python
import numpy as np

# Generate 'num_samples' samples from the Exponential distribution
# Each sample has 'sample_size_clt' data points
# The scale parameter for np.random.exponential is 1/lambda
samples_from_exp = [np.random.exponential(scale=1/lambda_param, size=sample_size_clt) for _ in range(num_samples)]

print(f"Generated {len(samples_from_exp)} samples.")
print(f"Each sample contains {len(samples_from_exp[0])} data points.")
print(f"First 5 data points of the first sample: {samples_from_exp[0][:5]}")
```

**Reasoning**:
Now that the samples are generated, I will calculate the mean for each of these 1000 samples and store them in a list for further analysis and plotting.



```python
sample_means = [np.mean(sample) for sample in samples_from_exp]

print(f"Calculated {len(sample_means)} sample means.")
print(f"First 5 sample means: {sample_means[:5]}")
```

**Reasoning**:
To represent the original Exponential distribution for plotting, I will generate a single large sample from the Exponential distribution using the defined `lambda_param`.



```python
import numpy as np

# Generate a single large sample to represent the original Exponential distribution
# This will be used for plotting the original distribution's histogram
# Using a large number like 100,000 for this 'population' sample to approximate the true distribution
original_population_sample = np.random.exponential(scale=1/lambda_param, size=100000)

print(f"Generated a single large sample of size {len(original_population_sample)} to represent the original Exponential distribution.")
print(f"Mean of original population sample: {np.mean(original_population_sample):.4f}")
print(f"Standard deviation of original population sample: {np.std(original_population_sample):.4f}")
```

**Reasoning**:
To prepare for plotting the normal distribution curve, I will calculate the mean and standard deviation of the `sample_means` data.



```python
import numpy as np

# Calculate the mean and standard deviation of the sample means
mu_sample_means = np.mean(sample_means)
sigma_sample_means = np.std(sample_means)

print(f"Mean of sample means (mu_sample_means): {mu_sample_means:.4f}")
print(f"Standard deviation of sample means (sigma_sample_means): {sigma_sample_means:.4f}")
```

**Reasoning**:
To visualize the Central Limit Theorem, I will plot two histograms: one for the original exponential distribution using a large sample and another for the distribution of the sample means. I will also overlay a normal distribution curve on the histogram of sample means.



```python
import matplotlib.pyplot as plt
from scipy.stats import norm

# Create a figure with two subplots
plt.figure(figsize=(15, 6))

# Plot 1: Histogram of the Original Exponential Distribution (Population)
plt.subplot(1, 2, 1)
plt.hist(original_population_sample, bins=50, density=True, alpha=0.7, color='skyblue', edgecolor='black')
plt.title('Original Exponential Distribution (Population Sample)')
plt.xlabel('Value')
plt.ylabel('Density')
plt.grid(True, linestyle='--', alpha=0.6)

# Plot 2: Histogram of Sample Means
plt.subplot(1, 2, 2)

# Plot histogram of sample means
plt.hist(sample_means, bins=30, density=True, alpha=0.7, color='lightcoral', edgecolor='black', label='Sample Means Histogram')

# Overlay a normal distribution curve
xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 100)
p_normal = norm.pdf(x, mu_sample_means, sigma_sample_means)
plt.plot(x, p_normal, 'r', linewidth=2, label=f'Normal PDF (mu={mu_sample_means:.2f}, sigma={sigma_sample_means:.2f})')

plt.title('Distribution of Sample Means')
plt.xlabel('Sample Mean')
plt.ylabel('Density')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

# Question 9: Write a Python function to calculate and visualize the confidence interval for a sample mean.

**Calculate and Plot Confidence Intervals with Python:**

Create a Python function that calculates and plots a confidence interval for a given sample data and confidence level. The function should handle both cases: when the population standard deviation is known (using a Z-score) and when it's unknown (estimating from the sample and using a T-score). Demonstrate the function's utility by generating sample data and plotting the confidence intervals for both scenarios, ensuring the plots clearly show the sample mean and the upper and lower bounds of the interval. Finally, interpret the results, highlighting the differences in calculation and visualization based on the knowledge of the population standard deviation.

## Import Libraries and Define Function

### Subtask:
Import necessary Python libraries and define a function `calculate_and_plot_confidence_interval` that calculates and plots a confidence interval based on whether the population standard deviation is known or unknown.


**Reasoning**:
The subtask requires importing necessary libraries and defining a function `calculate_and_plot_confidence_interval`. This function will encapsulate the logic for calculating and plotting confidence intervals based on whether the population standard deviation is known or unknown, as detailed in the instructions.



```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

def calculate_and_plot_confidence_interval(sample_data, confidence_level, population_std_dev=None):
    """
    Calculates and plots a confidence interval for a population mean.

    Parameters:
    -----------
    sample_data : array_like
        A 1-D array or list of numerical data.
    confidence_level : float
        The desired confidence level (e.g., 0.95 for 95%).
    population_std_dev : float, optional
        The known population standard deviation. If not provided (None), a T-test
        will be used; otherwise, a Z-test will be used.
    """
    sample_mean = np.mean(sample_data)
    sample_size = len(sample_data)

    # Determine critical value and standard error based on known/unknown population_std_dev
    if population_std_dev is not None:
        # Z-test (Population Std Dev Known)
        sem = population_std_dev / np.sqrt(sample_size)
        # For a two-tailed test, alpha is split into two tails
        alpha = 1 - confidence_level
        critical_value = stats.norm.ppf(1 - alpha / 2)
        method = 'Z-test (Population Std Dev Known)'
    else:
        # T-test (Population Std Dev Unknown)
        sample_std = np.std(sample_data, ddof=1) # ddof=1 for unbiased sample standard deviation
        sem = sample_std / np.sqrt(sample_size)
        df = sample_size - 1  # Degrees of freedom
        # For a two-tailed test, alpha is split into two tails
        alpha = 1 - confidence_level
        critical_value = stats.t.ppf(1 - alpha / 2, df)
        method = 'T-test (Population Std Dev Unknown)'

    # Calculate Margin of Error (MoE)
    margin_of_error = critical_value * sem

    # Calculate Confidence Interval
    lower_bound = sample_mean - margin_of_error
    upper_bound = sample_mean + margin_of_error

    # Print results
    print(f"Sample Mean: {sample_mean:.4f}")
    print(f"Method Used: {method}")
    print(f"Confidence Level: {confidence_level*100:.0f}%")
    print(f"Margin of Error: {margin_of_error:.4f}")
    print(f"Confidence Interval: ({lower_bound:.4f}, {upper_bound:.4f})")

    # Plotting the Confidence Interval
    plt.figure(figsize=(10, 3))
    plt.hlines(y=0, xmin=lower_bound, xmax=upper_bound, linewidth=5, color='blue', label='Confidence Interval')
    plt.plot(sample_mean, 0, 'o', color='red', markersize=10, label='Sample Mean')

    plt.text(lower_bound, 0.1, f'LB: {lower_bound:.2f}', verticalalignment='bottom', horizontalalignment='center', color='blue')
    plt.text(upper_bound, 0.1, f'UB: {upper_bound:.2f}', verticalalignment='bottom', horizontalalignment='center', color='blue')
    plt.text(sample_mean, -0.1, f'Mean: {sample_mean:.2f}', verticalalignment='top', horizontalalignment='center', color='red')

    plt.title(f'{confidence_level*100:.0f}% Confidence Interval ({method})')
    plt.xlabel('Value')
    plt.yticks([]) # Hide y-axis ticks
    plt.xlim(sample_mean - margin_of_error * 1.5, sample_mean + margin_of_error * 1.5) # Adjust x-axis limits
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.legend()
    plt.show()
```

# Question 10: Perform a Chi-square goodness-of-fit test using Python to compare observed and expected distributions, and explain the outcome.

### **1. Purpose of the Chi-square Goodness-of-Fit Test**

The Chi-square goodness-of-fit test is a statistical hypothesis test used to determine whether a sample distribution matches a population distribution. In other words, it tests if observed categorical data significantly differ from what would be expected under a hypothesized distribution.

It helps answer questions like: "Are the observed frequencies of categories consistent with the theoretically expected frequencies?"

### **2. Assumptions of the Chi-square Goodness-of-Fit Test**

For the Chi-square goodness-of-fit test to be valid, the following assumptions should be met:

*   **Categorical Data**: The data should be in categories (e.g., colors, preferences, types).
*   **Independence**: The observations must be independent of each other. Each observation falls into only one category.
*   **Expected Frequencies**: Expected frequencies for each category should be at least 5. If some categories have expected frequencies less than 5, they might need to be combined with other categories to meet this assumption.
*   **Random Sampling**: The data should be collected from a random sample of the population.

### **3. Example Scenario: Candy Colors**

Let's imagine a candy company claims that its bags of mixed candies contain 30% red, 20% yellow, 20% green, 15% orange, and 15% blue candies.

A curious consumer buys a large bag and counts the candies, obtaining the following observed frequencies:

*   **Red**: 60
*   **Yellow**: 45
*   **Green**: 38
*   **Orange**: 27
*   **Blue**: 30

**Total candies observed: 200**

We want to perform a Chi-square goodness-of-fit test to see if the observed distribution of candy colors significantly differs from the company's claimed distribution at a significance level (α) of 0.05.

### **4. Define Hypotheses and Significance Level**

*   **Null Hypothesis (H₀)**: The observed distribution of candy colors does not significantly differ from the company's claimed distribution (i.e., the observed proportions are equal to the expected proportions).
*   **Alternative Hypothesis (H₁)**: The observed distribution of candy colors significantly differs from the company's claimed distribution (i.e., at least one observed proportion is different from its expected proportion).

**Significance Level (α)**: 0.05

```python
import numpy as np
from scipy import stats

# Observed frequencies
observed_frequencies = np.array([60, 45, 38, 27, 30])

# Expected proportions (from the company's claim)
expected_proportions = np.array([0.30, 0.20, 0.20, 0.15, 0.15])

# Total number of observations
total_observations = np.sum(observed_frequencies)

# Calculate expected frequencies
expected_frequencies = total_observations * expected_proportions

# Significance level
alpha = 0.05

print(f"Observed Frequencies: {observed_frequencies}")
print(f"Expected Proportions: {expected_proportions}")
print(f"Total Observations: {total_observations}")
print(f"Calculated Expected Frequencies: {expected_frequencies}")
print(f"Significance Level (alpha): {alpha}")
```

### **5. Calculate the Chi-square Statistic and P-value**

We will use `scipy.stats.chisquare` to perform the test. This function returns both the Chi-square test statistic and the p-value.

```python
# Perform the Chi-square goodness-of-fit test
chi2_statistic, p_value = stats.chisquare(f_obs=observed_frequencies, f_exp=expected_frequencies)

print(f"Chi-square Statistic: {chi2_statistic:.4f}")
print(f"P-value: {p_value:.4f}")
```

### **6. Interpret the Result**

Compare the P-value to the significance level ($\alpha$).

*   If P-value $\le$ $\alpha$: Reject the null hypothesis.
*   If P-value $> \alpha$: Fail to reject the null hypothesis.

```python
print(f"Significance Level (alpha): {alpha}")
print(f"P-value: {p_value:.4f}")

if p_value <= alpha:
    print("\nDecision: Reject the Null Hypothesis (H0).")
    print(f"Conclusion: There is sufficient evidence at the {alpha*100}% significance level to conclude that the observed distribution of candy colors is significantly different from the company's claimed distribution.")
else:
    print("\nDecision: Fail to Reject the Null Hypothesis (H0).")
    print(f"Conclusion: There is NOT sufficient evidence at the {alpha*100}% significance level to conclude that the observed distribution of candy colors is significantly different from the company's claimed distribution. The observed differences could be due to random chance.")
```

### **7. Summary of the Chi-square Goodness-of-Fit Test Steps**

1.  **State Hypotheses**: Formulate the null (H₀) and alternative (H₁) hypotheses.
2.  **Set Significance Level ($\alpha$)**: Choose a threshold for statistical significance (e.g., 0.05).
3.  **Collect Data**: Gather observed frequencies for each category.
4.  **Calculate Expected Frequencies**: Determine the expected frequencies based on the hypothesized population distribution.
5.  **Calculate Chi-square Statistic**: Compute the Chi-square test statistic.
6.  **Calculate P-value**: Find the probability of observing a Chi-square statistic as extreme as, or more extreme than, the calculated one, assuming H₀ is true.
7.  **Make a Decision**: Compare the P-value with $\alpha$.
8.  **Formulate a Conclusion**: Interpret the decision in the context of the problem.
